```python
import pandas as pd
import re
import string
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
df = pd.read_csv(
    "/content/IMDB Dataset - IMDB Dataset.csv",
    encoding='utf-8',
    engine='python',
    on_bad_lines='skip'
)

# Take top 50 rows
df = df.head(50)
```

```python
# -----------------------------------
# Text Cleaning Functions
# -----------------------------------

# Lowercase
df['review'] = df['review'].str.lower()
```

```python
# Remove HTML tags
import re
def remove_html_tags(text):
  Pattern=re.compile('<.*?>')
  return Pattern.sub(r'',text)
```

```python
text="<html><body><p> m=Movie1 </p> Click here to <a heref='http://goggle.com'>download/a"
```

```python
remove_html_tags(text)
```

```python
df['review'] = df['review'].apply(remove_html_tags)
```

```python
# Remove URLs
def remove_url(text):
    pattern = re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub('', text)
```

```python
text4='for notebook click http://www.deture.com/delta/note1234acc'
```

```python
string.punctuation
```

```python
# Remove punctuation
exclude = string.punctuation
```

```python
def remove_punc(text):
  for char in exclude:
    text=text.replace(char,'')
    return text
```

```python
text3='string.with.punctuation?'
```

```python
remove_punc(text3)
```

```python
df['review']=df['review'].apply(remove_punc)
```

```python
df
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf=TfidfVectorizer()
tfidf.fit_transform(df['review']).toarray()
```

```python
print(tfidf.idf_)
print(tfidf.get_feature_names_out())
```
