```python
#NORMAL ENTITY RECOGNIZTION(NER) is an nlp technique used to find important names and information in
# text such as people,places,organization,dates and amounts  and label then with their correct
# type.

# in short:
# NER helps a machine understand "wha","what", and "where" from text
# Rohit Sharma lives in Mumbai.

# why do we use ner
# we use NER to convert unstructured text into meaningful,struture data.
# Main Reason:
# Automatically extract important information
# Readuce manual work in reading documents
# Help machine understand context,not just keywords.
# make text usable for analytics , Ml models ,and databases
```

```python
import spacy
```

```python
#load english nlp model
nlp=spacy.load("en_core_web_sm")
```

```python
text="barak obama was born in in hawaii and become a president of the united sate"
doc=nlp(text)
for ent in doc.ents:
  print(ent.text,"->",ent.label)
```

```python
import pandas as pd
import re

# Sample Resume Texts
resumes = [
    """This is Rahul Verma.
    Previously employed at TCS and Accenture.
    Total experience is 5 years.
    Skilled in SQL, Power BI, and Data Analysis.
    Location: Mumbai.
    Email: rahul.verma@outlook.com""",

    """This is Priya Sharma.
    Previously employed at Infosys and Wipro.
    Total experience is 3 years.
    Skilled in Python, Machine Learning, and SQL.
    Location: Pune.
    Email: priya.sharma@gmail.com""",

    """This is Aman Gupta.
    Previously employed at Capgemini and IBM.
    Total experience is 6 years.
    Skilled in Excel, Tableau, and SQL.
    Location: Delhi.
    Email: aman.gupta@yahoo.com""",

    """This is Sneha Patil.
    Previously employed at Cognizant and Deloitte.
    Total experience is 4 years.
    Skilled in Deep Learning, NLP, and Python.
    Location: Hyderabad.
    Email: sneha.patil@outlook.com""",

    """This is Vikas Mehta.
    Previously employed at HCL and Tech Mahindra.
    Total experience is 7 years.
    Skilled in AWS, Data Engineering, and Spark.
    Location: Bangalore.
    Email: vikas.mehta@gmail.com"""
]

structured_data = []

for text in resumes:

    # Name

    name = re.search(r'This is (.*?)\.', text).group(1)

    # Companies
    companies = re.search(r'employed at (.*?)\.',text).group(1)

    # Experience
    experience = re.search(r'(\d+) years',text).group(1)

    # Skills
    skills = re.search(r'Skilled in (.*?)\.',text).group(1)

    # Location
    location = re.search(r'Location: (.*?)\.',text).group(1)

    # Email
    email = re.search(r'Email: (.*?)$',text.strip()).group(1)

    structured_data.append({
        "Name": name,
        "Companies": companies,
        "Experience": experience + " years",
        "Skills": skills,
        "Location": location,
        "Email": email
    })

# Create DataFrame
df = pd.DataFrame(structured_data)
df
```
